```python
import numpy as np
from scipy.stats import mode
from sklearn.mixture import GaussianMixture
from sklearn.metrics import adjusted_rand_score
```

```python
seed = 44
data_path = 'path_to_tetrominoes_data'
gmm_logs = 'path_to_GMM_logs/'
vae_logs = 'path_to_vae_logs/'
ldm_logs = 'path_to_ldm_logs/'
slot_attn_logs = 'path_to_slot_attention_logs/'
slot_vae_logs = 'path_to_slot_vae_logs/'
mixture_mpm_logs = 'path_to_mixture_decoder_logs/'
additive_mpm_logs = 'path_to_additive_decoder_logs/'
```

```python
mix_mpm_aris = []
for i in range(42, 47):
    arr = np.load(data_path, allow_pickle=True)['arr_0']
    log_dir = f'{mixture_mpm_logs}/beta=0.01-seed={i}'
    alphas = np.load(f'{log_dir}/predictions/alphas.npy')
    
    N = len(arr)
    np.random.seed(i)
    np.random.shuffle(arr)
    arr = arr[int(0.9*N):]
    N = len(arr)
    
    gt_labels = []
    gt_masks = []
    labels = [a['mask'] for a in arr]
    labels = np.array(labels)
    labels = labels.reshape(N, 4, -1)
    labels = np.argmax(labels, axis=1)
    for i in range(N):
        mode_i = int(mode(labels[i])[0])
        gt_labels.append([int(k-1) for k in labels[i] if k != mode_i])
        gt_masks.append([i for i,k in enumerate(labels[i]) if k != mode_i])

    pred = np.argmax(alphas, axis=1).reshape(N, -1)
    pred = [a[gt_masks[i]] for i,a in enumerate(pred)]
    aris = [adjusted_rand_score(gt_labels[i], pred[i]) for i in range(N)]
    mix_mpm_aris.append(100 * np.mean(aris))
```

```python
gmm_aris = []
for seed in range(42, 47):
    arr = np.load(data_path, allow_pickle=True)['arr_0']
    N = len(arr)
    np.random.seed(seed)
    np.random.shuffle(arr)
    arr = arr[int(0.9*N):]
    N = len(arr)
    
    gt_labels = []
    gt_masks = []
    labels = [a['mask'] for a in arr]
    labels = np.array(labels)
    labels = labels.reshape(N, 4, -1)
    labels = np.argmax(labels, axis=1)
    for i in range(N):
        mode_i = int(mode(labels[i])[0])
        gt_labels.append([int(k-1) for k in labels[i] if k != mode_i])
        gt_masks.append([i for i,k in enumerate(labels[i]) if k != mode_i])

    pred_gmm = []
    gmm = GaussianMixture(
        max_iter=100,
        n_components=4,
        covariance_type="full",
        init_params="kmeans",
        random_state=i,
    )
    print(f'Training GMMs on seed {seed}')
    for i in range(N):
        data = arr[i]['image'] / 255
        data = data.reshape(-1, 3)
        gmm.fit(data)
        pred = gmm.predict(data)
        pred_gmm.append(pred[gt_masks[i]])
    aris_gmm = [adjusted_rand_score(gt_labels[i], pred_gmm[i]) for i in range(N)]
    gmm_aris.append(100 * np.mean(aris_gmm))
```
